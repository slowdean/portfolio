# A/B Testing AnalysisStatistical analysis of experiment results.

```python
# Importimport pandas as pdimport numpy as npimport matplotlib.pyplot as pltfrom scipy import statsplt.style.use('seaborn-v0_8-whitegrid')
```

## 1. Load Data

```python
df = pd.read_csv('data/ab_test_data.csv')print('Shape:', df.shape)df.head()
```

## 2. Calculate Metrics

```python
stats = df.groupby('variant').agg({    user_id: 'count',    conversion: ['sum', 'mean']}).reset_index()stats.columns = ['variant', 'visitors', 'conversions', 'rate']stats['rate'] = stats['rate'] * 100print(stats)
```

## 3. Chi-Square Test

```python
contingency = pd.crosstab(df['variant'], df['conversion'])chi2, p_value, dof, expected = stats.chi2_contingency(contingency)print(f'Chi2: {chi2:.4f}, p-value: {p_value:.6f}')print('Significant!' if p_value < 0.05 else 'Not significant')
```

## 4. Results

```python
a_rate = stats[stats['variant']=='A']['rate'].values[0]b_rate = stats[stats['variant']=='B']['rate'].values[0]lift = (b_rate - a_rate) / a_rate * 100print(f'A: {a_rate:.1f}%, B: {b_rate:.1f}%')print(f'Lift: +{lift:.1f}%')
```

## 5. Visualization

```python
plt.figure(figsize=(10,5))plt.subplot(1,2,1)plt.bar(['A (Control)', 'B (Test)'], [a_rate, b_rate], color=['#e74c3c','#2ecc71'])plt.ylabel('Conversion Rate (%)')plt.title('Conversion Rate Comparison')plt.subplot(1,2,2)df.groupby('variant')['time_on_page_seconds'].mean().plot(kind='bar', color=['#e74c3c','#2ecc71'])plt.title('Avg Time on Page')plt.tight_layout()plt.show()
```
