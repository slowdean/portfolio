# A/B Testing AnalysisStatistical analysis of A/B test results using Chi-Square test.

```python
# Import librariesimport pandas as pdimport numpy as npimport matplotlib.pyplot as pltfrom scipy import statsplt.style.use("seaborn-v0_8-whitegrid")
```

## 1. Load Data

```python
df = pd.read_csv('data/ab_test_data.csv')print('Shape:', df.shape)df.head()
```

## 2. Calculate Metrics by Variant

```python
stats_df = df.groupby('variant').agg({    user_id: 'count',    conversion: ['sum', 'mean']}).reset_index()stats_df.columns = ['variant', 'visitors', 'conversions', 'rate']stats_df['rate'] = stats_df['rate'] * 100print(stats_df)
```

## 3. Chi-Square Test

```python
contingency = pd.crosstab(df['variant'], df['conversion'])chi2, p_value, dof, expected = stats.chi2_contingency(contingency)print(f'Chi2: {chi2:.4f}')print(f'p-value: {p_value:.6f}')print('Significant!' if p_value < 0.05 else 'Not significant')
```

## 4. Results Summary

```python
a_rate = stats_df[stats_df['variant']=='A']['rate'].values[0]b_rate = stats_df[stats_df['variant']=='B']['rate'].values[0]lift = (b_rate - a_rate) / a_rate * 100print(f'Variant A (Control): {a_rate:.1f}%')print(f'Variant B (Test): {b_rate:.1f}%')print(f'Relative Lift: +{lift:.1f}%')
```

## 5. Visualization

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 5))axes[0].bar(['A (Control)', 'B (Test)'], [a_rate, b_rate], color=['#e74c3c','#2ecc71'])axes[0].set_ylabel('Conversion Rate (%)')axes[0].set_title('Conversion Rate Comparison')time_by_variant = df.groupby('variant')['time_on_page_seconds'].mean()axes[1].bar(['A (Control)', 'B (Test)'], time_by_variant.values, color=['#e74c3c','#2ecc71'])axes[1].set_ylabel('Avg Time on Page (s)')axes[1].set_title('Time on Page')plt.tight_layout()plt.show()
```
